```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('dataset_fraud/fraud.csv')

print(df.head())
```

```python
import seaborn as sns
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report,confusion_matrix
from sklearn.metrics import average_precision_score, f1_score
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.pipeline import Pipeline
```

```python
# 1. Set up the Imputer to fill in the blanks
imputer = IterativeImputer(max_iter=10, random_state=42)


rf_base = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)

# 3. Set up a Random Forest that is mathematically forced to care about the 10% fraud
# 'balanced' automatically makes a mistake on a fraud case 9x more painful for the AI
rf_weighted = RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42, n_jobs=-1)

# 4. Create the Pipelines (First fill the blanks, THEN train the model)
pipe_base = Pipeline([('imputer', imputer), ('classifier', rf_base)])
pipe_weighted = Pipeline([('imputer', imputer), ('classifier', rf_weighted)])

# Prepare a dictionary to hold our test scores
results = {"Baseline": [], "Class_Weights": [], "Manual_Undersample": []}
```

```python
# Assuming your main table is called 'df'

# 1. Define 'y' (The Answer Key)
# Replace 'fraud_label' with whatever your actual target column is named (e.g., 'is_fraud', 'Class', 'label')
y = df['is_fraud'] 

# 2. Define 'X' (The Clues)
# We drop the answer key column so the AI doesn't cheat by looking at the answers!
X = df.drop(columns=['is_fraud']) 

print("X and y successfully created!")
print(f"X has {X.shape[1]} clues. y has {len(y)} answers.")
```

```python
# Create the 5 splits, keeping the 10% fraud ratio intact
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for train_idx, val_idx in skf.split(X, y):
    # Separate the 'Study Material' (train) from the 'Final Exam' (val)
    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]
    
    # --- STRATEGY 1: The Baseline ---
    pipe_base.fit(X_train, y_train)
    y_probs_base = pipe_base.predict_proba(X_val)[:, 1]
    # average_precision_score is our PR-AUC metric
    results["Baseline"].append(average_precision_score(y_val, y_probs_base))
    
    # --- STRATEGY 2: Class Weights ---
    pipe_weighted.fit(X_train, y_train)
    y_probs_weight = pipe_weighted.predict_proba(X_val)[:, 1]
    results["Class_Weights"].append(average_precision_score(y_val, y_probs_weight))
```

```python
# --- STRATEGY 3: Manual Undersampling ---
    # 1. Find the scammers and the normal customers in the TRAINING data
train_fraud = X_train[y_train == 1]
train_legit = X_train[y_train == 0]
    
    # 2. Randomly select just enough normal customers to match the number of scammers
train_legit_downsampled = train_legit.sample(n=len(train_fraud), random_state=42)
    
    # 3. Stick them back together to make a new, perfectly balanced 50/50 training set
X_train_down = pd.concat([train_fraud, train_legit_downsampled])
y_train_down = pd.Series([1]*len(train_fraud) + [0]*len(train_legit_downsampled))
    
    # 4. Train the basic model on this new balanced dataset
pipe_base.fit(X_train_down, y_train_down)
y_probs_down = pipe_base.predict_proba(X_val)[:, 1]
results["Manual_Undersample"].append(average_precision_score(y_val, y_probs_down))
```

```python
print("Final PR-AUC Scores (Higher is Better, Baseline ~0.10):")
print("-" * 50)
for name, scores in results.items():
    print(f"{name:<20} | Mean PR-AUC: {np.mean(scores):.4f}")
```
