```python
import pandas as pd
import numpy as np
```

```python
train_df=pd.read_csv("./dataset_titatinc/train.csv")

display(train_df.head())
```

```python
test_df=pd.read_csv("./dataset_titatinc/test.csv")
display(test_df.head())
```

```python
features=["Pclass","Sex","SibSp","Parch"]

X_train=pd.get_dummies(train_df[features])
X_test=pd.get_dummies(test_df[features])

y_train=train_df["Survived"]

display(y_train.head())
```

```python
from sklearn.ensemble import RandomForestClassifier

# Initialize the model
model = RandomForestClassifier(n_estimators=100, random_state=1, max_depth=5)

# Train the model (Notice there is only one set of parentheses here!)
model.fit(X_train, y_train)

# Make predictions
predictions = model.predict(X_test)

# Package the predictions into a Kaggle-ready format
output = pd.DataFrame({
    'PassengerId': test_df['PassengerId'], 
    'Survived': predictions
})

# Save the final file
output.to_csv('titanic_submission.csv', index=False)
print("Success! Your submission file is saved.")
```
